```python
def multiply_for(a, b):
    result = 0
    for i in range(b):
        result = result + a
    return result

print("For loop: 7 x 6 =", multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    result = 0
    count = 0

    while count < b:
        result = result + a
        count = count + 1

    return result

print("While loop: 7 x 6 =", multiply_while(7, 6))
```

```python
print("3 x 5:", multiply_for(3, 5), multiply_while(3, 5))
print("7 x 6:", multiply_for(7, 6), multiply_while(7, 6))
print("10 x 4:", multiply_for(10, 4), multiply_while(10, 4))
```

The while loop stops when count becomes equal to b. The count increases by 1 each time the loop runs, so the loop cannot continue forever.
